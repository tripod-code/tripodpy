# Quick start

This script runs the default simulation and plots the resulting dust size distribution at 0.1 Myr.

The results are saved to the `data` folder in the working directory and should run in about 30s 

```python
from tripodpy import Simulation
from tripodpy import plot
```

```python
sim = Simulation()
sim.initialize()
sim.run()
```

```python
plot.panel(sim,show_limits=True,ir=20)
```
